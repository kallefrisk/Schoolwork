# Finding the Confidence Intervalls (CI)

## a. Beta(4,5)

```python
from scipy import stats

alpha, beta = 4, 5
lower, higher = stats.beta.ppf([0.025, 0.975], alpha, beta)

print(f"A 95% credible interval for π with π|y ~ Beta(4,5) is ({lower:.2f},{higher:.2f})")
```

## b. Beta(4.5)

```python
from scipy import stats

alpha, beta = 4, 5
lower, higher = stats.beta.ppf([0.1, 0.9], alpha, beta)

print(f"A 80% credible interval for π with π|y ~ Beta(4,5) is ({lower:.2f},{higher:.2f})")
```

## c. Gamma(1,8)

```python
from scipy import stats

s, r = 1, 8
lower, higher = stats.gamma.ppf([0.025, 0.975], s, scale=1/r)

print(f"A 95% credible interval for π with π|y ~ Gamma(1,8) is ({lower:.2f},{higher:.2f})")
```
