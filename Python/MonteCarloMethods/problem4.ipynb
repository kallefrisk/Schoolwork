### Simulation of a continuous random variable

Consider the random variable $X$ with distribution function

$F(x) = 1 − (1 − x)^{5}$ for $x \in [0, 1].$

a) Implement the inverse transform method to simulate values of X.

```python
import numpy as np

def transform(x):
	return 1 - (x)**(1/5)

U = np.random.uniform(0, 1)

random = transform(U)

print(random)
```

b) Implement the rejection method to simulate values of X.

```python
import numpy as np

def f(x):
    return 5*(1-x)**4

def g(x):
    return 1

while True:
    Y = np.random.uniform(0, 1)
    U = np.random.uniform(0, 1)
    
    M = 5
    
    if U <= f(Y)/(M * g(Y)):
        X = Y
        break

print(X)
```
