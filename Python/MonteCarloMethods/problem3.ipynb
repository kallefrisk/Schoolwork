### Simulation of a discrete random variable

Write a simulation algorithm for
generating a sequence of $100$ independent Bernoulli random variables, each with parameter $p = 0.24$. The
simulation algorithm should be based on the idea of generating a geometric random variable explained in
Examples 4d and 4e in the book. Use this algorithm to implement a Monte Carlo method estimating the
mean number of random numbers that are needed to generate $100$ independent Bernoulli random variables
in this way.

```python
import numpy as np

def geometric(p):
	q = 1 - p
	U = np.random.uniform(0, 1)
	return (np.log(U)/np.log(q) + 1) // 1

def bernoulli(size, p):
	array = np.zeros(size)
	index = int(geometric(p))
	if index == size:
		array[size-1] = 1
	elif index < size:
		array[index-1] = 1
		if size-index != 0:
			array = np.append(array[:index], bernoulli(size-index, p))
	return array


p = 0.24
q = 1 - p

simulations = 10000
size = 100
num_randoms = 0

for _ in range(simulations):
	trial = bernoulli(size, p)
	num_randoms += np.sum(trial)
	if trial[-1] == 0:
		num_randoms += 1

average_randoms_required = num_randoms/simulations

print(f"The average number of uniform random variables needed for 100 bernoulli random variables is: {average_randoms_required}.")
```
